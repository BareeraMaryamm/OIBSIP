```python
#list of column names that contain numbers
num_columns = ['quantity', 'unit_price', 'discount_pct', 'sales_amount']

#Run aggregation
stats = data[num_columns].agg(['mean', 'median', lambda x: x.mode()[0], 'std'])

#Rename index rows directly
stats.index = ['mean', 'median', 'mode', 'std']


stats.round(2)
```

</div>

```python
# Convert transaction_date to datetime format
data['transaction_date'] = pd.to_datetime(data['transaction_date'])

# Set transaction_date as index
df_time = data.set_index('transaction_date').sort_index()
```

<p style="color: #000; 
          font-size: 21px; 
          font-weight: bold; 
          text-align: left siide;">
    MONTHLY SALES ANALYSIS
</p>

```python
# Calculate Monthly Total Sales

monthly_sales = df_time['sales_amount'].resample('ME').sum().reset_index()
monthly_sales['Month'] = monthly_sales['transaction_date'].dt.to_period('M').astype(str)
monthly_sales.head()
```

```python
# Monthly Sales Line Chart

plt.figure(figsize=(10,5))
plt.plot(monthly_sales['transaction_date'], monthly_sales['sales_amount'], marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.grid(True)
plt.show()
```

<p style="color: #000; 
          font-size: 21px; 
          font-weight: bold; 
          text-align: left siide;">
    QUARTERLY SALES ANALYSIS
</p>

```python
# Calculate Quarterly Total Sales

quarterly_sales = df_time['sales_amount'].resample('QE').sum().reset_index()
quarterly_sales['Quarter'] = quarterly_sales['transaction_date'].dt.to_period('Q').astype(str)
quarterly_sales.head()
```

```python
# Quarterly Sales Line Chart

plt.figure(figsize=(8,4))
plt.plot(quarterly_sales['Quarter'], quarterly_sales['sales_amount'], marker='s', color='green')
plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales ($)")
plt.grid(True)
plt.show()
```

    1. <b style="color: #6ee7b7;">Revenue Concentration in Premium Categories:</b> Beauty, Books and Electronics drive the majority of revenue (over 3M each). Business should focus on premium inventory and upselling in these high-value categories rather than pushing volume in low-performing categories like Footwear.<br>
    2. <b style="color: #6ee7b7;">Age-Agnostic Demand - Universal Targeting Strategy:</b> Sales variance across age groups is less than 2% (4.11M to 4.17M), with 55+ being the top contributor at 4.17M. This indicates product appeal is universal, so marketing budget should be equally distributed instead of targeting a specific age bracket.<br>
    3. <b style="color: #6ee7b7;">Pricing Power is the Primary Growth Lever:</b> A strong correlation of 0.64 between unit_price and sales_amount proves that revenue is price-driven, not volume-driven. Strategy should prioritize value-based pricing and high-ticket items over high-quantity low-price sales.<br>
    4. <b style="color: #6ee7b7;">Discount Inefficiency:</b> Discount has a negligible negative correlation of -0.09 with sales_amount, indicating current discounting is eroding margins without driving incremental revenue. Recommend A/B testing discounts or shifting budget to product quality and branding.<br>
    5. <b style="color: #6ee7b7;">Volume Does Not Equal Value:</b> Very weak correlation of 0.14 between quantity and sales_amount suggests that operational focus on just increasing units sold is ineffective. Focus should shift from quantity to Average Order Value (AOV) optimization.
